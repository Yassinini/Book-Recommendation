## Book by name

```python
def search(val, typ):
    out = d[typ].str.contains(val, case=False, na=False) # used AI help for this line
    out = d[out]

    if len(out) ==0 :
        return "Book not found"
    
    elif len(out) ==1 :
        return out
   
    else:
        y = input(f"{out[['Name', 'Author']]} \n please choose a book by index")
        return out.iloc[int(y)]
```

```python
search("The great gatsby" , "Name")
```

## Book by specifics

```python
def specific(d):
    specific = input("Insert number followed by space then the specific available. \n 1. Name \n 2. Author \n 3. User Rating \n 4. Price \n 5. Genre \n ONLY ONE SPECIFIC ")
    
    num, val = specific.split(" ", 1)
    num = int(num)

    match num:
        case 1:
            return search(val, "Name")
        case 2:
            return search(val, "Author")
        case 3:
            return search(val, "User Rating")
        case 4:
            return search(val, "Price")
        case 5:
            return search(val, "Genre")
        case _:
            return "Invalid option"
```

```python
specific(d)
```

## Book recommend

```python
def searchtype(typ):
    if typ == 'Price':
        out = d.sort_values(by = typ, ascending=True)
    else:
        out = d.sort_values(by = typ, ascending=False)

    return out
```

```python
def recommend():
    r = int(input("Recommend by: \n 1. Rating \n 2. Price \n 3. Reviews \n 4. Year \n 5. Genre"))
    match r:
        case 1:
            return searchtype("User Rating")
        case 2:
            return searchtype("Price")
        case 3:
            return searchtype("Reviews")
        case 4:
            return searchtype("Year")
        case 5:
            return searchtype("Genre")
        case _:
            return "Invalid option"
```
