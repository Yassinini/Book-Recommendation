## Book by name

```python
def search(val, typ):
    matches = d[typ].astype(str).str.contains(val, case=False, na=False)
    out = d[matches]

    if len(out) == 0:
        return "Book not found"
    
    elif len(out) == 1:
        return out
   
    else:
        y = input(f"{out[['Name', 'Author']]} \n please choose a book by index")
        return out.iloc[int(y)]
```

```python
search("The great gatsby" , "Name")
```

## Book by specifics

```python
def specific(d):
    specific = input("Insert number followed by space then the specific available. \n 1. Name \n 2. Author \n 3. User Rating \n 4. Price \n 5. Genre \n ONLY ONE SPECIFIC ")
    
    num, val = specific.split(" ", 1)
    num = int(num)

    match num:
        case 1:
            return search(val, "Name")
        case 2:
            return search(val, "Author")
        case 3:
            return search(val, "User Rating")
        case 4:
            return search(val, "Price")
        case 5:
            return search(val, "Genre")
        case _:
            return "Invalid option"
```
